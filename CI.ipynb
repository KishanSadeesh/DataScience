```python
import pandas as pd
```

```python
data = pd.read_csv("Uber.csv")
data
```

```python
d = data['trips'][:30]
d
```

```python
import numpy as np
from scipy import stats

# 95% Confidence Interval automatically
ci = stats.t.interval(
    confidence=0.95,
    df=len(d) - 1,
    loc=np.mean(d),
    scale=stats.sem(d)
)

print("95% Confidence Interval:")
print("Lower:", ci[0])
print("Upper:", ci[1])
```

```python
population_std = 3

ci = stats.norm.interval(
    0.95,
    loc=np.mean(d),
    scale=population_std / np.sqrt(len(d))
)

print("Number of data:", len(d))
print("Mean:", np.mean(d))

print("95% Confidence Interval:")
print("Lower:", ci[0])
print("Upper:", ci[1])
```

```python
#1. 50 students, mean=165, SD=10; 95% CI
student=50
mean=165
sd=10
ci=stats.norm.interval(
    0.95,
    loc=mean,
    scale=sd/np.sqrt(student)
)
print("95% Confidence Interval:")
print("Lower:",ci[0])
print("Upper:",ci[1])
```

```python
#2. Bottle fill mean=500ml, SD=8, n=40; 99% CI
population_std = 8

ci = stats.norm.interval(
    0.99,
    loc=500,
    scale=population_std / 40
)

print("95% Confidence Interval:")
print("Lower:", ci[0])
print("Upper:", ci[1])
```

```python
#4. Restaurant rating mean=4.2, SD=0.5, n=80; 90% CI.
population_std = 0.5
mean = 4.2
n = 80
ci = stats.norm.interval(
    0.90,
    loc=mean,
    scale=population_std / n
)

print("95% Confidence Interval:")
print("Lower:", ci[0])
print("Upper:", ci[1])
```
