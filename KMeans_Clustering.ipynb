```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/JangirSumit/kmeans-clustering/refs/heads/master/driver-data.csv")
data
```

```python
X = data[['mean_dist_day','mean_over_speed_perc']]
X
```

```python
data.isnull().sum()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X)

for i in range(len(X)):
    print(f"Record {i+1} -> Cluster {clusters[i]}")
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    X["mean_dist_day"],
    X["mean_over_speed_perc"],
    c=clusters
)

plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    s=300,
    marker="X"
)

plt.xlabel("Fresh")
plt.ylabel("Milk")
plt.title("K-Means Driving Dataset")

plt.grid(True)

plt.show()
```
