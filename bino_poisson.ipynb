```python
#Poisson Distribution
#1. Website gets 6 visits/min; probability exactly 10 next minute.
import numpy as np
x = np.arange(3,15)
print(x)
from scipy.stats import poisson
lamda = 6
k = 10
ans = poisson.pmf(k,lamda)
print("poisson for k : ",ans)
result = poisson.pmf(x,lamda)
print("Array poisson distribution : ",result)
```

```python

import matplotlib.pyplot as plt
```

```python
plt.bar(x,result)
plt.xlabel("No of visitors")
plt.ylabel("Probability")
plt.show()
```

```python
#2. Bank gets 3 customers/5 min; probability 5 in next 5 min.
lamda = (3/5)*5
k = 5
x = np.arange(1,10)
print(x)
ans = poisson.pmf(k,lamda)
print("poisson for k : ",ans)
result = poisson.pmf(x,lamda)
print("Array poisson distribution : ",result)
```

```python
plt.bar(x,result)
plt.xlabel("No of customers")
plt.ylabel("Probability")
plt.show()
```

```python
#3. Hospital gets 10 patients/hr; probability >12 next hour.
lamda = 10
k = 12
x = np.arange(9,20)
print(x)
ans = poisson.pmf(k,lamda)
print("poisson for k : ",ans)
result = poisson.pmf(x,lamda)
print("Array poisson distribution : ",result)
```

```python
plt.bar(x,result,color="lightblue")
plt.xlabel("No of patients")
plt.ylabel("Probability")
plt.show()
```

```python
#2. Student solves 60% problems; probability solves at least 3 of 5.
n = 5
p = 0.60
k = 3
p = binom.pmf(k,n,p)
print(p)
```

```python
x = np.arange(0,5)
pmf_values = 1 - binom.pmf(x, n, p)
plt.bar(x, pmf_values, color="orange", edgecolor="black", label="Binomial PMF")
plt.bar(k, p, color="lightgreen", edgecolor="black", label=f"k = {k}")

plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title(f"Binomial distribution - PMF")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```
