```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("credit_risk_dataset.csv")
data
```

```python
data.isnull().sum()
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

```python
x = data[['loan_percent_income']]
x
```

```python
y = data['loan_status']
y
```

```python
y.unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
model = LogisticRegression()
model.fit(x,y)
```

```python
# salary = int(input("Enter the salary: "))
loan = float(input("Enter loan_percentage from salary score in percent: "))
loan_percent = loan/100
prediction = model.predict([[loan_percent]])
if prediction[0] == 1:
    print("\nRESULT : CLAIM")
else:
    print("\nRESULT : NOT CLAIM")
```

```python
x_lines = np.linspace(data['loan_percent_income'].min(),data['loan_percent_income'].max(),300).reshape(-1,1)
y_prob = model.predict(x_lines)
y_prob
```

```python
plt.scatter(data['loan_percent_income'],data['loan_status'],color="green")
plt.plot(x_lines,y_prob,color="red")
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv("Crop_recommendation.csv")
data
```

```python
x=data[['temperature','humidity','rainfall']]#you should take multiple columns for more accuracy
x
```

```python
y=data['label']
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=43)
```

```python
model=KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
```

```python
temperature=float(input("Enter temperature : "))
humidity=float(input("Enter humidity: "))
rainfall=float(input("Enter rainfall: "))
user_input=[[temperature,humidity,rainfall]]
print(model.predict(user_input))
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(
        model, x, y,scoring='accuracy'
    )
print("Average Accuracy =", round(scores.mean()*100,2),"%")
```
