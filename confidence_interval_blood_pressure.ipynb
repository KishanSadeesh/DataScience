```python
cal_t = np.array([10.26,10.61,11.19,10.85,16.97,16.55,12.53,10.79,11.15,14.56,8.47,8.75,11.26,8.99,16.55,11.98,10.96,16.42,12.53,9.08,16.91,11.36,9.38,11.77,14.56,11.69,15.53,9.69,11.07,16.91,10.14,13.63,17.17])
```

```python
len(cal_t)
```

```python
#T-distribution
ci = stats.t.interval(
    confidence = 0.95,
    df = len(cal_t) - 1,
    loc = np.mean(cal_t),
    scale = stats.sem(cal_t)
)

print("Lower limit : " + str(ci[0]))
print("Higher limit : " + str(ci[1]))
```

```python
cal_ran = np.array([10.79,11.15,16.91,11.36,9.38,11.77,14.56,11.69,15.53,9.69,11.07,16.91,10.14,13.63,17.17])
```

```python
len(cal_ran)
```

```python
pop_standard = np.std(cal_ran)
pop_standard
```

```python
#Z-distribution
ci = stats.norm.interval(
    0.95,
    loc = np.mean(cal_ran),
    scale = pop_standard / np.sqrt(len(cal_ran))
)

print("Lower limit : " + str(ci[0]))
print("Higher limit : " + str(ci[1]))
```
