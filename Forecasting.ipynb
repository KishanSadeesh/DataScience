```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/kalyanipatange/Waste-Food-Management-Analysis-/main/food_waste_dataset.csv")
data
```

```python
data['Date'] = pd.to_datetime(data["Date"])
data['Date']
```

```python
data = data.sort_values("Date")
data
```

```python
data.isnull().sum()
```

```python
input = data['Waste_Quantity']
input
```

```python
from statsmodels.tsa.arima.model import ARIMA

# 1. Aggregate the waste quantity daily to ensure unique dates
daily_data = data.groupby('Date')['Waste_Quantity'].sum().asfreq('D', fill_value=0)

model = ARIMA(daily_data, order=(5,1,0))
model_fit = model.fit()

forecast = model_fit.forecast(steps=10)
print(forecast)
```
