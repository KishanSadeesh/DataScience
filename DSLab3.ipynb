```python
import numpy as np
from scipy.stats import kurtosis as k
```

```python
data = pd.read_csv("batter.csv")
data
```

```python
avg = data['avg'][1:50].to_numpy()
avg
```

```python
type(avg)
```

```python
kurto=k(avg,fisher=True)
kurto1=k(avg,fisher=False)
meanval = np.mean(avg)
maxval = np.max(avg)
minval = np.min(avg)
print(kurto)
if kurto>0:
    print("LeptoKurtic")
    if(maxval-meanval)>(meanval-minval):
        print("big profits")
    else:
        print("big loss")
elif(kurto<0):
    print("low tail less distribution and low profit/loss")
else:
    print("normal distribution and normal profit/loss")
```

```python
print(kurto)
if kurto < 3:
    print("Platykurtic")
elif kurto > 3:
    print("Leptokurtic")
else:
    print("Mesokurtic")
```
