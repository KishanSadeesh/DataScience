```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv("House_Rent_Dataset.csv")
data
```

```python
x = data[['Size']]
x
```

```python
y = data['Rent']
y
```

```python
# Correcting the train_test_split unpacking order
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)

# Fitting the model
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
size = int(input("Enter size : "))
predicted_price = model.predict([[size]])
print("Predicted Price for size : ",predicted_price)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    data["Size"],
    data["Rent"]
)

plt.plot(
    data["Size"],
    model.predict(x),
    linewidth=3
)

plt.xlabel("Flat Size")
plt.ylabel("House Rent")
plt.title("House Rent based on size")

plt.grid(True)

plt.show()
```

```python
accuracy = r2_score(y_test, y_pred)
print(accuracy*100)
```

```python
data = pd.read_csv("ToyotaCorolla-selected-columns.csv")
data
```

```python
data.isnull().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
x1 = data['Age_08_04']
x1
```

```python
x2 = data['KM']
x2
```

```python
from numpy import column_stack

x = column_stack((x1,x2))
x
```

```python
y = data['Price']
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy = r2_score(y_test, y_pred)
print(accuracy*100)
```

```python
age = int(input("Enter age of car : "))
km = int(input("Enter total km driven by car : "))
predicted_price = model.predict(column_stack((age,km)))
print("Predicted Price for car details provided : ",predicted_price)
```

```python
fig = plt.figure(figsize=(10, 7))

ax = fig.add_subplot(
    111,
    projection='3d'
)

ax.scatter(
    data['Age_08_04'],
    data['Price'],
    data['KM']
)

ax.set_xlabel("Age of Car")
ax.set_ylabel("Price of Car")
ax.set_zlabel("Km driven")
plt.title(
    "Car Price prediction using km and age of used cars"
)

plt.show()
```

```python
data = pd.read_csv("Churn_Modelling-selected-columns.csv")
data
```

```python
print(data.isnull().sum())
```

```python
data.drop_duplicates(inplace=True)
print(data.shape)
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
le = LabelEncoder()

data["Exited"] = le.fit_transform(data["Exited"])
```

```python
x = data[['CreditScore', 'Age']]
y = data['Exited']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state=42,stratify=y)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
model = LogisticRegression(class_weight='balanced', random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy : ",round(accuracy*100,4),"%")
```
