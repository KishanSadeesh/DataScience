```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("data.csv")
data
```

```python
a = data["Income"]
a
```

```python
from scipy.stats import skew

result = skew(a)
print(result)

if result > 0:
    print("Positive Skew -- Right Tail")
elif result < 0:
    print("Negative Skew -- Left Tail")
else:
    print("Symmetric Skew")
```

```python
import matplotlib.pyplot as mp
mp.hist(a,color="blue")
mp.xlabel("Income")
mp.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.hist(a, color='skyblue', bins=10)
plt.title("Profit Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.stats  import binom
import numpy as np
n=3
p=0.5
k=2
prob=binom.pmf(k,n,p)
print(prob)
```

```python
x = np.arange(0, n + 1)
pmf_values = binom.pmf(x, n, p)
plt.bar(x, pmf_values, color="orange", edgecolor="black", label="Binomial PMF")
plt.bar(k, prob, color="lightgreen", edgecolor="black", label=f"k = {k}")

plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title(f"Binomial distribution - PMF")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
n=12
p=0.6
k=10
p=binom.cdf(k,n,p)
print(p)
```

```python
x = np.arange(0, n + 1)
pmf_values = binom.cdf(x, n, p)
plt.bar(x, pmf_values, color="orange", edgecolor="black", label="Binomial CDF")
plt.bar(k, p, color="lightgreen", edgecolor="black", label=f"k = {k}")

plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title(f"Binomial distribution - CDF")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```
